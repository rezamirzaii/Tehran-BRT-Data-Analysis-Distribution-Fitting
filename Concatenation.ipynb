```python
today = datetime.date.today()
counter = 0
for index, row in station.iterrows():
    for i, j in inside.iterrows():
      if (row['Station'] == j['Station']) and (row['position'] == j['position']) and (row['Date'] == j['Date']):
        diff = datetime.datetime(1300,1,1)
        s_date = datetime.datetime.combine(today, row['Bus arrival time'])
        i_date = datetime.datetime.combine(today, j['Bus arrival time'])
        if row['Bus arrival time'] > j['Bus arrival time']:
          diff = s_date - i_date
        else:
          diff = i_date - s_date

        if diff.total_seconds() < 60:
          counter += 1
          inside.drop(i,inplace=True)
print(counter)
```
